# Dicts & sets

**Creating a dict**

```python
music_data = {
    'Jose': 'Jimi Hendrix',
    'Iva': 'Midori Takada',
    'Misha': 'BTS',
    'Pat': 'Igor Stravinsky'
}
music_data
```

**Access by key**

```python
music_data['Misha']
```

**Missing key**

Accessing a missing key raises a KeyError.

```python
music_data['Reza']
```

**Safe access**

Use .get() to avoid a KeyError on a missing key.

```python
music_data.get('Reza')
music_data.get('Reza', 'J.S. Bach')
```

**Add / update a key**

```python
music_data['Valerie'] = 'Johnny Cash'
music_data['Iva'] = 'Steve Reich'
music_data
```

**Delete a key**

Use del to remove a key from a dict.

```python
del music_data['Pat']
music_data
```

**pop() removes and returns**

```python
music_data.pop('Valerie')
```

**Membership**

The in operator checks keys, not values.

```python
'Jose' in music_data
'Jimi Hendrix' in music_data
```

**Iterate keys**

Iterating a dict by default gives its keys.

```python
for key in music_data:
    print(key)
```

**Iterate values**

```python
for value in music_data.values():
    print(value)
```

**Iterate key, value pairs**

```python
for key, value in music_data.items():
    print(key, value)
```

**len(), keys(), values(), items()**

```python
len(music_data)
list(music_data.keys())
list(music_data.values())
list(music_data.items())
```

**Dict comprehension**

```python
lengths = {name: len(song) for name, song in music_data.items()}
lengths
```

**Nested dicts**

```python
students = {
    'Misha': {'programme': 'MSc FinTech', 'grade': 72},
    'Iva': {'programme': 'MSc BA', 'grade': 68}
}
students['Misha']['grade']
```

**Sets**

**Creating a set**

```python
berries = {'Raspberry', 'Cloudberry', 'Lingonberry'}
berries
```

**Sets remove duplicates**

```python
orders = ['Tiramisu', 'Fruit', 'Fruit', 'Tiramisu', 'Apple pie']
set(orders)
```

**Sets are unordered**

```python
for berry in berries:
    print(berry)
```

**add() / remove()**

```python
berries.add('Blueberry')
berries.remove('Lingonberry')
berries
```

**remove() errors if missing; discard() doesn't**

```python
berries.remove('Kiwi')
```

```python
berries.discard('Kiwi')
```

**Membership**

The in operator is fast because sets use hashing.

```python
'Blueberry' in berries
```

**Empty set**

Use set() for an empty set, since {} makes a dict.

```python
fruit = set()
fruit.add('Apple')
fruit
```

**Set operations**

Union, intersection, difference, and symmetric difference.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
a | b
a & b
a - b
a ^ b
```

**Why sets are fast**

Hashing gives O(1) average lookup, versus O(n) for a list.
