# Web scraping & APIs

**Parse HTML with BeautifulSoup**

```python
from bs4 import BeautifulSoup
html = '<p>Hello, world!</p> <p>Hello again.</p>'
soup = BeautifulSoup(html, 'html.parser')
soup.text
```

**find_all**

Returns every matching tag.

```python
paragraphs = soup.find_all('p')
paragraphs
type(paragraphs)
```

**Access one tag like a list item**

```python
first = paragraphs[0]
first.name
first.text
type(first)
```

**requests**

Downloads a real page from the web.

```python
import requests
r = requests.get('http://www.example.com')
r.ok
```

**r.text = raw HTML string**

```python
r.text
```

**Parse it, find links**

```python
soup = BeautifulSoup(r.text, 'html.parser')
links = soup.find_all('a')
links
```

**Get the href attribute**

```python
first_link = links[0]
first_link.text
first_link.attrs
first_link['href']
```

**Real example**

The Guardian's best breakfasts page.

```python
r = requests.get("https://www.theguardian.com/lifeandstyle/2017/jan/15/50-best-breakfasts-uk")
soup = BeautifulSoup(r.text, 'html.parser')
```

**Inspect the page HTML first**

Show dev tools live.

**Find all restaurant headers**

```python
results = soup.find_all('h2')
results[0:2]
```

**Extract name + link from one result**

```python
link = results[0].find('a')
name = link.text
url = link['href']
name, url
```

**Loop over all results**

Build a list of dicts, one per result.

```python
result_list = []
for result in results:
    link = result.find('a')
    if link:
        result_list.append({'Restaurant': link.text, 'Web page': link['href']})
result_list[:3]
```

**Write to CSV**

```python
import csv
with open('best_breakfast_guardian.csv', 'w', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['Restaurant', 'Web page'])
    writer.writeheader()
    writer.writerows(result_list)
```

**Scraping is messy**

A layout change breaks everything.

**APIs**

**Prefer an API when one exists**

APIs return clean, structured JSON.

```python
url = "https://api.datausa.io/tesseract/data.jsonrecords?cube=acs_yg_total_population_5&measures=Population&drilldowns=State,Year"
response = requests.get(url)
data = response.json()
data.keys()
```

**Convert to a DataFrame**

```python
import pandas as pd
df = pd.DataFrame(data['data'])
df.head()
```

**API > scraping**

No HTML parsing, and no risk of layout changes breaking your code.
