**Functions**

**How to write comments**

```python
# This is a comment: Python ignores everything after the #
x = 5  # comments can also follow code on the same line
x
```

**Built-in functions**

Python includes many built-in functions, like abs and max.

```python
y = abs(-5)
x = max(y, 3)
x
```

**Before we define our own**

This is how we'd build an exclamation manually, before writing a function.

```python
name = 'Reza'
exclamation = name + '!'
exclamation
```

**f-strings vs concatenation**

f-strings are safer since they handle type conversion for you.

```python
age = 30

# 'Age: ' + age  # TypeError (can't add a str and an int directly)
'Age: ' + str(age)  # concatenation needs an explicit conversion

f'Age: {age}'  # an f-string converts automatically
```

**Defining a function**

This one takes a single parameter.

```python
def exclaim(name):
    excl = name + '!'
    return excl
```

**Code after return never runs**

Return immediately ends the function.

```python
def example():
    return 'done'
    print('this line never runs')

example()
```

**Docstrings**

The """ """ right after def documents what the function does.

```python
def square(x):
    """Return the square of x."""
    return x * x

help(square)
```

**Calling it**

```python
exclaim('Reza')
```

**Any number of parameters**

Functions can take zero, one, or many parameters.

```python
def say_hello():
    return 'Hello!'

say_hello()
```

**Nested and chained calls**

The result of one call becomes the input to another.

```python
exclaim(exclaim('Reza'))
```

**Two parameters**

A function can take multiple arguments.

```python
def greet(name, greeting):
    base_greeting = greeting + ' ' + name
    excl = exclaim(base_greeting)
    return excl
```

**Calling it**

```python
greet('Reza', 'Hello')
```

**What if we forget to return?**

```python
def greet_no_return(name, greeting):
    base_greeting = greeting + ' ' + name
    excl = exclaim(base_greeting)
    # no return, so Python returns None by default
```

**Watch what happens**

```python
result = greet_no_return('Reza', 'Hello')
result + 'hey'
```

**More than one output?**

Functions can have more than one output.

```python
def min_max(a, b):
    smallest = min(a, b)
    largest = max(a, b)
    return smallest, largest

low, high = min_max(3, 9)
low, high
```

**Default values**

Parameters can have default values, so callers can leave them out.

```python
def greet_default(name, greeting='Hello'):
    return greeting + ' ' + name + '!'

greet_default('Reza')
```

```python
greet_default('Reza', 'Hi')
```
