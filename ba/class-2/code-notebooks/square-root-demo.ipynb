### Square roots

**Initial guess**

Start by guessing that the square root is half of x.

```python
x = 9
guess = x / 2
print(guess)
```

**One iteration**

Average the guess with x/guess.

```python
ratio = x / guess
average = (guess + ratio) / 2
guess = average
print(guess)
```

**Repeat again**

Average the guess with x/guess again.

```python
ratio = x / guess
average = (guess + ratio) / 2
guess = average
print(guess)
```

**Keep going**

Each repetition gets the guess closer to the true answer.

```python
ratio = x / guess
average = (guess + ratio) / 2
guess = average
print(guess)
```

**While loop**

Turn the repetition above into a while loop.

**Reset and repeat**

Keep averaging until the guess is close enough.

```python
# how to know where to stop iterating when approximating the square root?
x = 10000
guess = x / 2
print(guess)

tolerance = 0.001
error = abs(guess*guess - x)

while error >= tolerance:
    ratio = x / guess
    average = (guess + ratio) / 2
    guess = average
    error = abs(guess*guess - x)
    print(guess)
```

**Wrap it up**

Put the while loop inside a function.

```python
def square_root(x):
    guess = x / 2
    tolerance = 0.001
    error = abs(guess*guess - x)
    while error >= tolerance:
        ratio = x / guess
        average = (guess + ratio) / 2
        guess = average
        error = abs(guess*guess - x)
    return guess
```

**Try it**

```python
square_root(10000)
```
