### Debugging in VS Code

**Objective: cumulative sum**

Write a function that takes a list of numbers and returns a new list of *running* totals: each position is the sum of everything up to and including that one.

For `[1, 3, 2, 4]`:
- `1` (sum of items from index 0 to 0)
- `(1) + 3 = 4` (sum of items from index 0 to 1)
- `(1 + 3) + 2 = 6` (sum of items from index 0 to 2)
- `(1 + 3 + 2) + 4 = 10` (sum of items from index 0 to 3)

So `csum([1, 3, 2, 4])` should return `[1, 4, 6, 10]`.

**Step 1: keep a running sum**

Before building a list, let's just track the running total as we go.

```python
numbers = [1, 3, 2, 4]
cumul_sum = 0
for value in numbers:
    cumul_sum = cumul_sum + value
    print(cumul_sum)
```

**Step 2: building a list in a loop**

A different skill: start with an empty list, and append to it as you go. For example, a list of squares of another list:

```python
other_numbers = [2, 5, 3]
squares = []
for n in other_numbers:
    squares.append(n ** 2)
squares
```

**Combining the two**

Track the running sum, and append it to a results list at each step.

```python
numbers = [1, 3, 2, 4]
cumul_sum = 0
results = []
for value in numbers:
    cumul_sum = cumul_sum + value
    results.append(cumul_sum)
results
```

**Wrap it in a function**

Same logic, packaged so we can reuse it on any list.

```python
def csum(numbers):
    cumul_sum = 0
    results = []
    for value in numbers:
        cumul_sum = cumul_sum + value
        results.append(cumul_sum)
    return results

numbers = [1, 3, 2, 4]  # expect [1, 4, 6, 10]
csum(numbers)
```

**A second attempt: csum2**

Let's try writing it differently, using slicing instead of a running total.

```python
def csum2(values):
    results = []
    for i in range(0, len(values)-1):
        # calculate sum of values up to i
        results.append(sum(values[0:i]))
    return results

numbers = [1, 3, 2, 4]  # expect [1, 4, 6, 10]
res = csum2(numbers)
print(res)
```

**Something's wrong**

The output doesn't match [1, 4, 6, 10]. Time to debug.

**Print debugging**

Add print statements to see what's happening at each step.

```python
def csum2(values):
    results = []
    for i in range(0, len(values)-1):
        print('i =', i)
        slice_of_values = values[0:i]
        print('slice:', slice_of_values)
        sum_of_values = sum(slice_of_values)
        results.append(sum_of_values)
    return results

numbers = [1, 3, 2, 4]  # expect [1, 4, 6, 10]
csum2(numbers)
```

**Step through with the debugger**

Let's look at the same buggy code again, this time using VS Code's debugger instead of prints.

**Set a breakpoint**

Click just left of the line number inside csum2, then run with the debugger (F5) instead of just running the file.

**Step through the loop**

Use Step Over (F10) to move line by line, and watch i and results update in the Variables pane.

**Inspect a variable**

Hover over i or values[0:i], or check the Variables pane, to see its value at each step.

**Spot the bug**

range(0, len(values)-1) stops one index too early, and values[0:i] leaves out index i itself.

**The fix**

range(0, len(values)) and values[0:i+1] now cover every index.

```python
def csum2(values):
    results = []
    for i in range(0, len(values)):
        results.append(sum(values[0:i+1]))
    return results

numbers = [1, 3, 2, 4]  # expect [1, 4, 6, 10]
csum2(numbers)
```

**Try it**

Confirm the fix on a different list.

```python
csum2([5, 1, 2])
```

**Teaching moment: assert**

assert checks a condition and raises an error immediately if it's false: a quick way to catch bugs instead of silently getting a wrong answer.

```python
assert csum2([1, 3, 2, 4]) == [1, 4, 6, 10]
print('Passed!')
```

**What if it's wrong?**

assert raises an AssertionError immediately when the condition is false.

```python
assert csum2([1, 3, 2, 4]) == [1, 2, 3, 4]
```
