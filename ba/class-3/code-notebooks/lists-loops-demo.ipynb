### Lists and for loops

**While loop recap**

A quick reminder, before we compare it to a for loop.

```python
i = 0
while i < 4:
    print(i)
    i = i + 1
```

**For loop**

Does the same job, more concisely.

```python
for i in range(4):
    print(i)
```

**range with start, stop, step**

range(start, end, step) counts up between two values.

```python
for i in range(5, 11, 2):
    print(i)
```

**Negative step**

The step can be negative too, to count down.

```python
for i in range(10, 7, -1):
    print(i)
```

**Lists**

An ordered collection, written with square brackets.

```python
prices = [5.2, 4.9, 5.1, 5.6, 7.0]
prices
```

**Creating and growing a list**

Start empty, then add items with .append().

```python
names = []
names.append('Ruth')
names.append('Mo')
names
```

**len() and indexing**

Count items, or access one by position, starting at 0.

```python
print(len(names))
print(names[0])
```

**Slicing**

Grab a range of items: names[start:stop].

```python
names[0:2]
```

**Changing an item**

Lists are mutable: assign to an index to replace a value.

```python
names[1] = 'Olivia'
names
```

**Looping over a list**

for item in list works directly, no index needed.

```python
for name in names:
    print('Hello ' + name)
```

**Looping by index**

Useful when you need the position too.

```python
for index in range(len(names)):
    print('Hi ' + names[index] + ', your index is ' + str(index))
```

**enumerate()**

Gives you index and value together, like .items() does for dictionaries.

```python
for index, name in enumerate(names):
    print(index, name)
```

**A bit more list syntax**

A few more common operations, worth knowing.

**Membership: in**

Check whether a value is present, without looping yourself.

```python
'Ruth' in names
```

**insert()**

Add an item at a specific position.

```python
names.insert(0, 'Ana')
names
```

**Negative indexing**

Index -1 is the last item, -2 the second-last, and so on.

```python
names[-1]
```

**remove()**

Removes the first matching value.

```python
names.remove('Olivia')
names
```

**pop()**

Removes and returns the last item.

```python
names.pop()
```
