**Let's build a more useful Monster**

We'll add attributes and methods one at a time.

**Add __init__**

This special method runs automatically when we call Monster(...), to set up a new instance. Methods with double underscores on both sides like this are called dunder methods (short for "double underscore"), or magic methods: Python calls them for you in specific situations, rather than you calling them directly.

```python
class Monster:
    
    def __init__(self, name, health):
        self.name = name
        self.health = health
```

```python
pika = Monster('Pikachu', 35)
bulb = Monster('Bulbasaur', 55)
char = Monster('Charzard', 22)
```

**Add say_hello**

self refers to the specific instance a method is called on, so each Monster uses its own name.

```python
class Monster:

    def __init__(self, name, health):
        self.name = name
        self.health = health

    def say_hello(self):
        print(self.name + ' says hello!')
```

```python
pika = Monster('Pikachu', 35)
bulb = Monster('Bulbasaur', 55)

pika.say_hello()
bulb.say_hello()
```

**A method call is just a function call**

pika.say_hello() is really Monster.say_hello(pika): self is filled in automatically.

```python
Monster.say_hello(pika)
```

**Add hurt**

hurt() changes state on that one instance.

```python
class Monster:

    def __init__(self, name, health):
        self.name = name
        self.health = health

    def say_hello(self):
        print(self.name + ' says hello!')

    def hurt(self, damage):
        self.health = self.health - damage
        if self.health <= 0:
            print(self.name + ' is knocked out!')
```

```python
pika = Monster('Pikachu', 35)
bulb = Monster('Bulbasaur', 55)

pika.hurt(20)
print(pika.health)
bulb.hurt(60)
```

**Add __str__**

Defines what print() shows for an instance, instead of a generic memory address.

```python
class Monster:

    def __init__(self, name, health):
        self.name = name
        self.health = health

    def say_hello(self):
        print(self.name + ' says hello!')

    def hurt(self, damage):
        self.health = self.health - damage
        if self.health <= 0:
            print(self.name + ' is knocked out!')

    def __str__(self):
        return 'A monster called ' + self.name
```

```python
pika = Monster('Pikachu', 35)
print(pika)
```

**Another dunder method: __eq__**

By default, == checks whether two variables point to the exact same instance.

```python
class Monster:
    """A simple game monster with a name and health."""

    def __init__(self, name, health):
        self.name = name
        self.health = health

    def say_hello(self):
        print(self.name + ' says hello!')

    def hurt(self, damage):
        self.health = self.health - damage
        if self.health <= 0:
            print(self.name + ' is knocked out!')

    def __str__(self):
        return 'A monster called ' + self.name

pika1 = Monster('Pikachu', 35)
pika2 = Monster('Pikachu', 35)
pika1 == pika2
```

**Define what == should mean**

Add __eq__ so two Monsters with the same name count as equal.

```python
class Monster:
    """A simple game monster with a name and health."""

    def __init__(self, name, health):
        self.name = name
        self.health = health

    def say_hello(self):
        print(self.name + ' says hello!')

    def hurt(self, damage):
        self.health = self.health - damage
        if self.health <= 0:
            print(self.name + ' is knocked out!')

    def __str__(self):
        return 'A monster called ' + self.name

    def __eq__(self, other):
        return self.name == other.name

pika1 = Monster('Pikachu', 35)
pika2 = Monster('Pikachu', 100)
pika1 == pika2
```

**Add __call__**

Define what happens when a Monster is called directly, like a function.

```python
class Monster:
    """A simple game monster with a name and health."""

    def __init__(self, name, health):
        self.name = name
        self.health = health

    def say_hello(self):
        print(self.name + ' says hello!')

    def hurt(self, damage):
        self.health = self.health - damage
        if self.health <= 0:
            print(self.name + ' is knocked out!')

    def __str__(self):
        return 'A monster called ' + self.name

    def __eq__(self, other):
        return self.name == other.name

    def __call__(self, damage):
        self.hurt(damage)

pika = Monster('Pikachu', 35)
```

```python
pika(20)
print(pika.health)
```
